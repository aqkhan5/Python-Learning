```python
# Attributes & Methods

# Attribute
  # Class Attribute:   A normal variable created inside a class is a class atribute and thats it.
  # Instance Attribute: An attribute created using an instance like self.name, self.age etc.

class Animal:
  a = 15     # Class Attribute
  def __init__ (self, name):
    self.name = name # object/instance attribute
  # Instance/Object Method: when you are creating a method and that method is capturing the location of the object ( usually using self ) that method is Instance method

  def hello(self):
    print(" Hello! how are you")


# Method
  # Class methods: Captures the location of class and its attributes you can use self or any other thing is that place
  @classmethod
  def details(cls):
    print(f"Its looking heavy!")

  # Static Method:  it didn't target the location of object or class.
  @staticmethod
  def voice():
    print("Roar, Roar, Roar! ")


obj = Animal("Lion")

print(obj.name)

obj.hello()

obj.details()
```

```python
# Inheritance
# In inheritance one class inherits the attributes and methods of the other class. There is a parent child relation between them.

class Animal:                # Parent Class
  def __init__(self, name):
    self.name = name
  def details(self):
    print(f"My name is {self.name}")

class Human(Animal):   # Child class
  pass


obj_1 = Animal("Lion")
obj_2 = Human("Hadi Khan")

print(obj_1.name)
print(obj_2.name)
obj_2.details()

# You child class object has all the powers to access the attributes and methods of parent class.
```

```python
# Multi-Level-Inheritance

class BagFactory:
  def __init__(self, material, zips, pockets):
    self.material = material
    self.zips = zips
    self.pockets = pockets

  def details(self):
    print("Your bag details are ")
    print(self.material)
    print(self.zips)
    print(self.pockets)

class Rebook(BagFactory):
  def __init__(self, material, zips, pockets, color):
    super().__init__(material, zips, pockets)
    self.color = color
  def details(self):
    print(self.color)
    return super().details()

class Campus(Rebook):
  def __init__(self, material, zips, pockets, color, size):
    super().__init__(material, zips, pockets, color)
    self.size = size
  def details(self):
    print(self.size)
    return super().details()

bag_1 = BagFactory("Leather", 3, 2)
bag_2 = Rebook("Polyester", 4, 2, "red")
bag_3 = Campus("Leather", 3, 2, "black", "Large")

bag_1.details()
bag_2.details()
bag_3.details()
```

```python
from os import name
# Multiple-inheritance

class Animal:
  def __init__(self, name):
    self.name = name

  def details(self):
    print(f"My name is {self.name}")
class Human:
  def __init__(self, id):
    self.id = id

  def details(self):
    print(f"My Id is {self.id}")

class Robot(Human, Animal):
  def __init__(self, id, name):
    Human.__init__(self, id)
    Animal.__init__(self, name)
robo = Robot(12, "Qadeer")
robo.details()
```

```python
# Dunder Methods
#  These are the special methods in python that start and end with double underscore, like __init__, __str__, __add__, etc.
#  They automatically get called when you perform certain action on an object.

class Animal:
  def __init__ (self, name):
    self.name = name
  def __str__ (self):
    return f"hello my name is {self.name}"
obj = Animal("Lion")
print(obj)

# Arithematic Dunder methods
class number:
  def __init__ (self, num):
    self.num = num
  def __add__ (self, other):
    return self.num + other.num
num_1 = number(10)
num_2 = number(20)
print(num_1 + num_2)
```

```python
# Decorator
# A decorator wrap a function to add extra behaviour without modifying its code
# Think of it like a gift wrapping, the gift remain same but now there is a wrapping around it.
def timer(func):
  def wrapping():
    print("...Starting.....")
    func()
    print("...Done....!")
  return wrapping

@timer
def greet():
  print("Hello!")
greet()
```

```python
# args  & kwargs
def addition(*args):
  s=0
  for i in args:
    s += i
  return s
print(addition(1,2,3,4,5,6,7,8,9))

#kwargs
def info(**kwargs):
  return kwargs
print(info(name = "Ali", age = 25, profession = "Data Engineer"))
```

```python
# args & kwargs together
def extragreeting(func):
  def wrapper(*args, **kwargs):
    print("hello from the NYC team")
    func(*args, **kwargs)
  return wrapper
@extragreeting
def addition (a, b, c):
  print(a + b +c )
addition(10, 20, 30)
```
